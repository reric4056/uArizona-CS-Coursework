```python
# Code for Simulation and Theoretical Framework
# Purpose: Verifying independence
import numpy as np

sampleSpace = np.arange(1, 7) # sample space for fair die: {1,2,3,4,5,6}

def event_from_set(outcomes):
    outcomes = np.array(sorted(outcomes))
    return lambda x: np.isin(x, outcomes)
    
def simulate_independence(A_set, B_set, n=1_000_000, seed=0):
    np.random.seed(seed)
    draws = np.random.randint(1, 7, size=n) # required by prompt
    
    A = event_from_set(A_set)(draws)
    B = event_from_set(B_set)(draws)
    
    pA = A.mean()
    pB = B.mean()
    pAandB = (A & B).mean()
    prod = pA * pB

    pAgivenB = A[B].mean()
    
    return pA, pB, pAandB, prod, pAgivenB
    
def theoretical_probs(A_set, B_set):
    A_set = set(A_set)
    B_set = set(B_set)
    S_set = set(sampleSpace.tolist())
    
    pA = len(A_set) / len(S_set)
    pB = len(B_set) / len(S_set)
    pAandB = len(A_set & B_set) / len(S_set)
    prod = pA * pB
    
    return pA, pB, pAandB, prod
    
def report(A_set, B_set, question):
    pA_s, pB_s, pAandB_s, prod_s, pAgivenB_s = simulate_independence(A_set, B_set)
    pA_t, pB_t, pAandB_t, prod_t = theoretical_probs(A_set, B_set)
    
    print(f"--- {question} ---")
    print("Simulation:")
    print(f"P(A) ≈ {pA_s:.6f}")
    print(f"P(B) ≈ {pB_s:.6f}")
    print(f"P(A∩B) ≈ {pAandB_s:.6f}")
    print(f"P(A)P(B) ≈ {prod_s:.6f}")
    print(f"Diff ≈ {pAandB_s - prod_s:.2e}")
    print(f"P(A|B) ≈ {pAgivenB_s:.6f}")
    
    print("\nTheoretical (exact):")
    print(f"P(A) = {pA_t:.6f}")
    print(f"P(B) = {pB_t:.6f}")
    print(f"P(A∩B) = {pAandB_t:.6f}")
    print(f"P(A)P(B) = {prod_t:.6f}")
    print("Independent?", pAandB_t == prod_t)
    print()
```

```python
report(A_set={2, 4, 6}, B_set={1, 2, 3, 4}, question="Q1")
```
