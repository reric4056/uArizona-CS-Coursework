(X, Y)^T ~ N(μ, Σ)
μ = (0 0)
Covariant Matrix: Σ : (1 / 0.6 // 0.6 / 1)

μ(x) = μ(y) = 0

σ(x) = σ(y) = 1

p = 0.06

```python
# Problem 5
import numpy as np
np.random.seed(0)

n = 500

mean_vector = np.array([0,0])

covariance_matrix = np.array([[1, 0.6], [0.6, 1]])

sample_data = np.random.multivariate_normal(mean_vector, covariance_matrix, size=n)

x_vals = sample_data[:, 0]
y_vals = sample_data[:, 1]

x_mean = np.mean(x_vals)
y_mean = np.mean(y_vals)

x_center = x_vals - x_mean
y_center = y_vals - y_mean

# Building using correlation formula
numerator = np.sum(x_center * y_center)
left_denominator = np.sum(x_center ** 2)
right_denominator = np.sum(y_center ** 2)

denominator = np.sqrt(left_denominator * right_denominator)

p = numerator / denominator

print(round(p, 3))
```

```python
# Problem 6
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)

m = 5000
n = 500

mean_vector = np.array([0,0])

covariance_matrix = np.array([[1, 0.6], [0.6, 1]])

estimates = np.zeros(m)

for i in range(m):
    sample_data = np.random.multivariate_normal(mean_vector, covariance_matrix, size=n)

    x_vals = sample_data[:, 0]
    y_vals = sample_data[:, 1]
    
    x_mean = np.mean(x_vals)
    y_mean = np.mean(y_vals)
    
    x_center = x_vals - x_mean
    y_center = y_vals - y_mean
    
    # Building using correlation formula
    numerator = np.sum(x_center * y_center)
    left_denominator = np.sum(x_center ** 2)
    right_denominator = np.sum(y_center ** 2)
    
    denominator = np.sqrt(left_denominator * right_denominator)
    
    estimates[i] = numerator / denominator

plt.hist(estimates, bins=30)
plt.xlabel("Estimated correlation")
plt.ylabel("Frequency")
plt.show()
```

```python
# Problem 7

true_correlation = 0.6

error_margin = estimates - true_correlation

margin_squared = error_margin ** 2

mse_estimate = np.mean(margin_squared)

print(round(mse_estimate, 6))
```
